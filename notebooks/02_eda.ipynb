# Phase 3B: Exploratory Data Analysis & Bivariate Relationships

**Project Title**: AI-Based Sales Conversion Probability Prediction  
**Dataset**: UCI Online Shoppers Purchasing Intention Dataset  
**Source**: `data/raw/online_shoppers_intention.csv`  


```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style="whitegrid")
df = pd.read_csv("../data/raw/online_shoppers_intention.csv")
df.head()
```

## 1. Conversion Rate Analysis Across Categorical Features

```python
for col in ["Month", "VisitorType", "Weekend", "OperatingSystems", "Region", "TrafficType"]:
    print(f"\n=== Conversion Rate by {col} ===")
    res = df.groupby(col)["Revenue"].agg(["count", "mean"]).rename(columns={"mean": "Conversion_Rate"})
    res["Conversion_Rate_Pct"] = res["Conversion_Rate"] * 100
    display(res)
```

## 2. Bivariate Numerical Feature Comparisons

```python
num_cols = df.select_dtypes(include=[np.number]).columns
print("=== MEAN NUMERICAL VALUES BY REVENUE CLASS ===")
display(df.groupby("Revenue")[num_cols].mean().T)
```
